```python
import tensorflow as tf
# 1. 가상의 데이터 설정
# 인코더가 처리한 입력 문장의 단어가 총 5개, 은닉 상태 크기가 32차원일 때
encoder_outputs = tf.random.normal([5, 32]) # shape: [5, 32]
print("encoder_outputs:\n", encoder_outputs)
# 현재 디코더가 가지고 있는 은닉 상태 (Decoder Hidden State)
decoder_state = tf.random.normal([1, 32]) # shape: [1, 32]
print("decoder_state:\n", decoder_state)
```

```python
print("--- 1. 어텐션 스코어(연관성 점수) 계산 ---")
# 디코더 상태와 인코더 각 단어들의 유사도를 구합니다. (내적 연산)
# [1, 32]와 [5, 32]의 전치 행렬([32, 5])을 곱해 [1, 5] 점수 산출
attention_scores = tf.matmul(decoder_state, encoder_outputs,
transpose_b=True)
print("점수 Shape:", attention_scores.shape)
print("점수:", attention_scores)
```

```python
print("\n--- 2. 소프트맥스를 통한 어텐션 가중치(확률) 계산 ---")
# 점수를 합이 1이 되는 확률 값으로 변환합니다.
attention_weights = tf.nn.softmax(attention_scores, axis=-1)
print("가중치 확률 (어디에 가장 집중하고 있을까요?):\n", attention_weights)
```

```python
print("\n--- 3. 동적 컨텍스트 벡터(Dynamic Context Vector) 생성 ---")
# 가중치([1, 5])를 인코더 출력([5, 32])에 곱해주어 최종 컨텍스트 벡터를 생성
context_vector = tf.matmul(attention_weights, encoder_outputs)
print("새로 만들어진 컨텍스트 벡터 Shape:", context_vector.shape)
print("컨텍스트 벡터 일부:\n", context_vector[0, :5])
```
