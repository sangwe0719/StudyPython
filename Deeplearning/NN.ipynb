```python
import numpy as np

x = np.array([0.5, 0.8, 0.2])
w = np.array([0.4, 0.7, 0.8])
d = 1
alpha = 0.5

for i in range (5):
    v = np.sum(w * x)
    y = v
    e = d - y
    dw = alpha * e * x
    w = w + dw

    print("Repeat", i + 1)
    print("e =", e)
    print("dw =", dw)
    print("w =", w)

   
```
