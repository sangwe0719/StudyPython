```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
# 1. 학습용 텍스트 데이터 준비
text = """
머신러닝과 딥러닝은 인공지능의 핵심 기술입니다.
인공지능은 우리 삶을 편리하게 만들어 줍니다.
딥러닝 모델 중 순환 신경망은 텍스트 처리에 유용합니다.
"""
# 2. 토큰화 (Tokenizer) 및 정수 인코딩
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
total_words = len(tokenizer.word_index) + 1 # 패딩을 위해 +1
print(f"총 단어 수: {total_words}")
```

```python
# 텍스트를 정수 시퀀스로 변환
tokenized_text = tokenizer.texts_to_sequences([text])[0]
print(f"정수 인코딩된 텍스트: {tokenized_text}")
# 3. 학습을 위한 n-gram 시퀀스 생성 (X: 입력 단어들, y: 다음 단어)
input_sequences = []
for i in range(1, len(tokenized_text)):
 n_gram_sequence = tokenized_text[:i+1]
 input_sequences.append(n_gram_sequence)
print(f"생성된 n-gram 시퀀스: {input_sequences}")
```

```python
# 패딩(Padding)을 통해 시퀀스 길이를 동일하게 맞춤
max_sequence_len = max([len(seq) for seq in input_sequences])
input_sequences = pad_sequences(input_sequences,
 maxlen=max_sequence_len, padding='pre')
print(f"패딩 후 시퀀스: {input_sequences}")
```

```python
# X(입력)와 y(정답 레이블) 분리
X = input_sequences[:, :-1]
y = input_sequences[:, -1]
print(f"X (입력): {X}")
print(f"y (정답 레이블): {y}")
```

```python
# y를 원-핫 인코딩(One-hot Encoding)으로 변환
y = tf.keras.utils.to_categorical(y, num_classes=total_words)
# 4. SimpleRNN 모델 구축
model = Sequential()
# Embedding 레이어: 단어 인덱스를 저차원의 밀집 벡터로 변환
model.add(Embedding(input_dim=total_words, output_dim=10,
 input_length=max_sequence_len-1))
# SimpleRNN 레이어: 순차적인 텍스트 패턴 학습
# 은닉 상태(Hidden State)의 차원 크기: 32
model.add(SimpleRNN(32))
model.add(Dense(total_words, activation='softmax'))
model.compile(loss='categorical_crossentropy', optimizer='adam',
 metrics=['accuracy'])
model.summary()
```

```python
# 5. 모델 학습
model.fit(X, y, epochs=500, verbose=0)
print("모델 학습 완료!")
model.summary()
```

```python
# 6. 간단한 텍스트 생성 테스트 함수
def generate_text(seed_text, next_words, model, max_sequence_len):
    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences([seed_text])[0]
        token_list = pad_sequences([token_list],
            maxlen=max_sequence_len-1, padding='pre')
        print(f"토큰화된 입력 시퀀스: {token_list}")
         
         # 예측 수행
        predicted = np.argmax(model.predict(token_list, verbose=0), axis=-1)
        print(f"예측된 다음 단어의 인덱스: {predicted}")
        
        # 예측된 정수를 다시 단어로 변환
        output_word = ""
        for word, index in tokenizer.word_index.items():
            if index == predicted:
                output_word = word
                break
        seed_text += " " + output_word
    return seed_text
```

```python
# 테스트: "딥러닝은"이라는 단어로 시작해 다음 3단어 생성하기
print(generate_text("딥러닝은", 3, model, max_sequence_len))
```
